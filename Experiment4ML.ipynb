```python
#Experiment 4 of Machine learning
#Import Necessary modules
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score

```

```python
#Generate sample data
X,y_true=make_blobs(n_samples=400,centers=4,cluster_std=0.60,random_state=42)

#Make dataframes
df =pd.DataFrame(X,columns = ["Features 1","Features 2"])
df["True Clusters"]=y_true

print(df.head())
```

```python
#Initialize and fit K-Means Model(K=4)
kmeans = KMeans(n_clusters=4,random_state=42,n_init="auto")
labels = kmeans.fit_predict(X)
centroids =kmeans.cluster_centers_
```

```python
silhouette_avg =silhouette_score(X,labels)

print("=== K-Means Clustering ===")
print(f"numbers of clusters(k):4")
print(f"Inertia(wcss):{kmeans.inertia_}")
print(f"Silhouette Score:{silhouette_avg:.2f}")
print(f"No.of iterations to converge:{kmeans.n_iter_}")

print("\n Cluster Centers:")
for i ,center in enumerate(centroids):
    print(f"Cluster {i}: [{center[0]:.2f},{center[1]:.2f}")
```

```python
new_points =np.array([[0,2],[3,1],[-3,-3]])
predicted_clusters = kmeans.predict(new_points)

print("\n=== Prediction foe new Data Points ===")
for point , cluster in zip(new_points,predicted_clusters):
    print(f"Point{point.tolist()} -> Cluster {cluster}")
```
